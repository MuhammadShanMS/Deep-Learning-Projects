```python
from tensorflow.keras.datasets import mnist
```

```python
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.fashion_mnist.load_data()
```

```python
from tensorflow.keras.utils import to_categorical
encoded_train_labels = to_categorical(train_labels)
encoded_test_labels = to_categorical(test_labels)
```

```python
print(train_labels[10])
encoded_train_labels[10]
```

```python
from tensorflow import keras
from tensorflow.keras import layers, Sequential
model = Sequential([
                    layers.Dense(512, activation="relu"),
                    layers.Dense(256, activation="relu"),
                    layers.Dense(10, activation="softmax")
                    ] )
```

```python
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy", metrics=["accuracy"])
# model.compile(optimizer="adam",loss="categorical_crossentropy", metrics=["accuracy"])
```

```python
train_images = train_images.reshape((60000, 784))
train_images = train_images.astype("float32") / 255
test_images = test_images.reshape((10000, 784))
test_images = test_images.astype("float32") / 255
```

```python
model.fit(train_images, train_labels, epochs=35, batch_size=128)
```

```python
model.evaluate(test_images, test_labels)
```

```python
import numpy as np

# Test images par predictions calculate karein
predictions = model.predict(test_images)

# Pehli image ki predicted class index
print("Predicted Class Index:", np.argmax(predictions[0]))
# Actual Label Check Karein
print("Actual Label:", test_labels[0])
```
